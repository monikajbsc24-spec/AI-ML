### 2. Load Tokenizer and Model

We'll load the `gpt2` tokenizer and model. The tokenizer will convert our text into numerical IDs that the model can understand, and the model will be the base for our fine-tuning.

```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

model_name = 'gpt2'
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

# Set the pad_token for GPT-2 tokenizer
# GPT-2 does not have a default pad token, which is needed for batching during training
tokenizer.pad_token = tokenizer.eos_token
model.config.pad_token_id = tokenizer.eos_token_id

print(f"Tokenizer and model '{model_name}' loaded successfully.")
```

### 3. Prepare Data for Training

We need to tokenize our dataset and format it for the causal language modeling task. The input to the model will be the combined `short_form -> expanded_form` string, and the model will learn to predict the next token in the sequence.

```python
from torch.utils.data import Dataset, DataLoader

class GenZDataset(Dataset):
    def __init__(self, texts, tokenizer, max_length=128):
        self.tokenizer = tokenizer
        self.input_ids = []
        self.attn_masks = []

        for text in texts:
            encodings = tokenizer(text, truncation=True, padding='max_length', max_length=max_length, return_tensors='pt')
            self.input_ids.append(encodings['input_ids'])
            self.attn_masks.append(encodings['attention_mask'])

    def __len__(self):
        return len(self.input_ids)

    def __getitem__(self, idx):
        return {
            'input_ids': self.input_ids[idx].squeeze(),
            'attention_mask': self.attn_masks[idx].squeeze(),
            'labels': self.input_ids[idx].squeeze() # For Causal LM, labels are typically the input_ids
        }

# Update to use the extended datasets
train_dataset = GenZDataset(train_extended_df['text'].tolist(), tokenizer)
eval_dataset = GenZDataset(eval_extended_df['text'].tolist(), tokenizer)

print(f"Training dataset size: {len(train_dataset)}")
print(f"Evaluation dataset size: {len(eval_dataset)}")
```

### 4. Fine-tune the Model

We'll use the `Trainer` API from the `transformers` library, which simplifies the fine-tuning process. We need to define `TrainingArguments` and then create a `Trainer` instance.

```python
from transformers import TrainingArguments, Trainer

# Define training arguments
training_args = TrainingArguments(
    output_dir='./results',
    num_train_epochs=3,           # number of training epochs
    per_device_train_batch_size=2, # batch size per device during training
    per_device_eval_batch_size=2,  # batch size for evaluation
    warmup_steps=10,             # number of warmup steps for learning rate scheduler
    weight_decay=0.01,            # strength of weight decay
    # logging_dir='./logs',         # DEPRECATED: Logs will now be automatically saved to output_dir/runs/...
    logging_steps=10,             # log every N updates steps
    eval_strategy="epoch",  # Evaluate at the end of each epoch
    save_strategy="epoch",        # Save the model checkpoint at the end of each epoch
    load_best_model_at_end=True,  # Load the best model at the end of training
    metric_for_best_model="eval_loss", # Metric to use to compare models
    report_to=["tensorboard"],     # Explicitly report metrics to TensorBoard
)

# Create a Trainer instance
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=eval_dataset,
    # tokenizer=tokenizer, # Removed this argument as it's no longer accepted
)

# Train the model
trainer.train()

print("Fine-tuning complete!")
```

### 6. Inspect Training Loss

During training, the `Trainer` saves logs (including loss values) in the directory specified by `logging_dir` (which is `./logs` in our case). We can use TensorBoard's `EventAccumulator` to read these logs and visualize the training progress.

```python
from tensorboard.backend.event_processing.event_accumulator import EventAccumulator
import matplotlib.pyplot as plt
import os
import re

# The Trainer typically saves logs in a 'runs' subdirectory within the output_dir
output_base_dir = './results'

if not os.path.exists(output_base_dir):
    print(f"Output directory '{output_base_dir}' not found. Please ensure training has completed and generated logs.")
else:
    # The Trainer creates 'runs/TIMESTAMP_FOLDER' inside the output_dir
    # We need to find the latest TIMESTAMP_FOLDER
    runs_dir_path = os.path.join(output_base_dir, 'runs')

    if not os.path.isdir(runs_dir_path):
        print(f"TensorBoard 'runs' directory not found at: {runs_dir_path}")
    else:
        timestamped_dirs = []
        for entry in os.listdir(runs_dir_path):
            full_path = os.path.join(runs_dir_path, entry)
            # Check if it's a directory and if it contains a typical TensorBoard event file
            # or matches a common timestamp pattern
            if os.path.isdir(full_path) and (re.match(r'^\d{4}-\d{2}-\d{2}_\d{2}-\d{2}-\d{2}', entry) or any('events.out.tfevents' in f for f in os.listdir(full_path))):
                timestamped_dirs.append(full_path)

        timestamped_dirs.sort(key=os.path.getmtime) # Sort by modification time to get the latest

        if not timestamped_dirs:
            print(f"No timestamped TensorBoard log directories found within '{runs_dir_path}'.")
        else:
            latest_log_dir = timestamped_dirs[-1]
            print(f"Loading logs from: {latest_log_dir}")

            # Initialize EventAccumulator
            # Setting size_guidance to 0 for scalars means it won't cap the number of items fetched
            event_acc = EventAccumulator(latest_log_dir, size_guidance={'scalars': 0})
            event_acc.Reload()

            # Print all available scalar tags for debugging
            print("\nAvailable scalar tags:", event_acc.Tags()['scalars'])

            # Extract scalar data for loss
            training_steps, training_loss = [], []
            eval_steps, eval_loss = [], []

            # Check for 'loss' or 'train/loss' for training loss
            if 'train/loss' in event_acc.Tags()['scalars']:
                for s in event_acc.Scalars('train/loss'):
                    training_steps.append(s.step)
                    training_loss.append(s.value)
            elif 'loss' in event_acc.Tags()['scalars']:
                for s in event_acc.Scalars('loss'):
                    training_steps.append(s.step)
                    training_loss.append(s.value)
            else:
                print("Training loss ('train/loss' or 'loss') not found in logs.")

            if 'eval/loss' in event_acc.Tags()['scalars']:
                for s in event_acc.Scalars('eval/loss'):
                    eval_steps.append(s.step)
                    eval_loss.append(s.value)
            else:
                print("Evaluation loss ('eval/loss') not found in logs.")

            if training_loss or eval_loss:
                plt.figure(figsize=(10, 6))
                if training_loss:
                    plt.plot(training_steps, training_loss, label='Training Loss')
                if eval_loss:
                    plt.plot(eval_steps, eval_loss, label='Evaluation Loss')

                plt.title('Training and Evaluation Loss Over Steps')
                plt.xlabel('Steps')
                plt.ylabel('Loss')
                plt.legend()
                plt.grid(True)
                plt.show()
            else:
                print("No loss data found to plot.")
```

### 5. Test the Fine-tuned Model

Now let's use the fine-tuned model to expand some Gen Z short forms and see how it performs.

```python
from transformers import pipeline

# Create a text generation pipeline with the fine-tuned model
fine_tuned_generator = pipeline('text-generation', model=model, tokenizer=tokenizer)

def expand_genz_slang(prompt_text, max_new_tokens=30):
    # GPT-2 is a causal language model, so we need to provide a complete prompt
    # that includes the short form and the '->' separator.
    generated_text = fine_tuned_generator(
        prompt_text,
        max_new_tokens=max_new_tokens,
        num_return_sequences=1,
        do_sample=True,
        temperature=0.7,
        pad_token_id=tokenizer.eos_token_id # Ensure pad_token_id is set
    )
    # Extract the generated text and try to isolate the expansion
    full_output = generated_text[0]['generated_text']
    try:
        # Assuming the format is 'SHORT -> EXPANDED'
        expansion = full_output.split('->')[-1].strip()
        return expansion
    except IndexError:
        return full_output # Return full output if splitting fails

# Test cases
short_form1 = "POV"
prompt1 = f"{short_form1} ->"
display(f"Input: '{short_form1}'")
display(f"Fine-tuned expansion: {expand_genz_slang(prompt1)}")

short_form2 = "IYKYK"
prompt2 = f"{short_form2} ->"
display(f"Input: '{short_form2}'")
display(f"Fine-tuned expansion: {expand_genz_slang(prompt2)}")

short_form3 = "Cap"
prompt3 = f"{short_form3} ->"
display(f"Input: '{short_form3}'")
display(f"Fine-tuned expansion: {expand_genz_slang(prompt3)}")

short_form4 = "FR"
prompt4 = f"{short_form4} ->"
display(f"Input: '{short_form4}'")
display(f"Fine-tuned expansion: {expand_genz_slang(prompt4)}")
```

```python
from torch.utils.data import Dataset, DataLoader

class GenZDataset(Dataset):
    def __init__(self, texts, tokenizer, max_length=128):
        self.tokenizer = tokenizer
        self.input_ids = []
        self.attn_masks = []

        for text in texts:
            encodings = tokenizer(text, truncation=True, padding='max_length', max_length=max_length, return_tensors='pt')
            self.input_ids.append(encodings['input_ids'])
            self.attn_masks.append(encodings['attention_mask'])

    def __len__(self):
        return len(self.input_ids)

    def __getitem__(self, idx):
        return {
            'input_ids': self.input_ids[idx].squeeze(),
            'attention_mask': self.attn_masks[idx].squeeze(),
            'labels': self.input_ids[idx].squeeze() # For Causal LM, labels are typically the input_ids
        }

# Update to use the extended datasets version 2
train_dataset = GenZDataset(train_extended_df_v2['text'].tolist(), tokenizer)
eval_dataset = GenZDataset(eval_extended_df_v2['text'].tolist(), tokenizer)

print(f"Training dataset size (v2): {len(train_dataset)}")
print(f"Evaluation dataset size (v2): {len(eval_dataset)}")
```

## Concept: Causal Language Modeling (CLM)

Causal Language Modeling (CLM) is a type of language modeling where the model is trained to predict the **next token in a sequence**, given all previous tokens. This is in contrast to Masked Language Modeling (MLM), where the model predicts masked tokens based on both left and right context.

Here's a breakdown of key aspects:

*   **Unidirectional Context:** In CLM, the model only looks at tokens that have *already occurred* in the sequence to predict the subsequent token. It cannot see future tokens. This makes the generation process causal, as each predicted token influences the prediction of the next.

*   **Generative Models:** Models trained with CLM are naturally good at text generation. When you provide a prompt, the model predicts the most probable next word, then uses that word (along with the original prompt) to predict the next, and so on, recursively generating new text.

*   **Example Application:** Our GPT-2 fine-tuning task is a perfect example of CLM. The model is given a sequence like "POV ->" and then predicts the word "Point", then "of", then "View", building the expansion token by token based on what it has generated so far.

*   **Training Objective:** During training, for a given input sequence `x1, x2, ..., xN`, the model tries to maximize the probability of `x_t` given `x1, ..., x_{t-1}` for all `t` from `2` to `N`. Essentially, it's learning to predict every token in the sequence based on its preceding context.

*   **Architectures:** Transformer decoder-only architectures (like GPT-2, GPT-3, Llama) are typically used for CLM, as their self-attention mechanisms are designed to prevent tokens from attending to future tokens in the sequence.
